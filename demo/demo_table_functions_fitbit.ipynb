```python
tbl=Table.read_table('https://raw.githubusercontent.com/lujiec2020/UMBC-IS296-Sp26/refs/heads/main/data/fitbit-sleep.csv')
tbl
```

```python
REM_sleep=tbl.column('Minutes REM Sleep')
deep_sleep=tbl.column('Minutes Deep Sleep')
awake = tbl.column('Minutes Awake')
total_sleep = tbl.column('Minutes Asleep')
```

```python
REM_percentage = REM_sleep/total_sleep * 100
REM_percentage
```

```python
SQI = (REM_sleep/total_sleep*100)*.4 + (deep_sleep/total_sleep*100)*.4 + (1-awake/total_sleep)*100*0.2
```

```python
import numpy as np
```

```python
np.mean(SQI)
```

```python
np.std(SQI)
```

```python
np.max(SQI)
```
